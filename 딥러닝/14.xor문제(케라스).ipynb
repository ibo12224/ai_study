## hidden layer층: relu

```python
myin  = Input(shape=(2,) ) # 입력층: input layer
dense1 = Dense( units=32, activation='relu')# w:2x32 b:32   #hidden layer
dense2 = Dense( units=16, activation='relu')# w:32x16 b: 16  #hidden layer
dense3 = Dense( units=1,  activation='sigmoid')# w:16x1 b: 1  # output layer
model = Sequential( [myin, dense1, dense2, dense3] )
model.compile( loss='binary_crossentropy', optimizer=Adam(0.1), metrics=['acc'] )
h = model.fit( x,y, epochs=100 )
```

### 퀴즈 : iris 데이터 딥러닝으로  구성해서 train, test 의 정확도 확인

```python
from sklearn.datasets import load_iris
from keras.utils import to_categorical
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
```

```python
iris = load_iris()
```

```python
x = iris['data']
y = to_categorical( iris['target'] )
```

```python
x_train, x_test, y_train,y_test = train_test_split(x,y, test_size=0.2,stratify=y)
```

```python
myin  = Input(shape=(4,) ) 
dense1 = Dense( units=32, activation='relu')
dense2 = Dense( units=16, activation='relu')
dense3 = Dense( units=3,  activation='softmax')
model = Sequential( [myin, dense1, dense2, dense3] )
model.compile( loss='categorical_crossentropy', optimizer=Adam(0.01), 
              metrics=['acc'] )
h = model.fit( x_train,y_train, epochs=500 )
```

```python
pred = model.predict( x_train ).argmax(axis=1)
accuracy_score( y_train.argmax(axis=1), pred )
```

```python
pred = model.predict( x_test ).argmax(axis=1)
accuracy_score( y_test.argmax(axis=1), pred )
```
