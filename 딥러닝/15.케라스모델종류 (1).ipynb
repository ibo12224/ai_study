```python
import tensorflow as tf
from keras.optimizers import Adam
from keras import Sequential, Model
from keras.layers import Dense, Input
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.array( [[0,0],[0,1],[1,0],[1,1]] , np.float32)
y = np.array( [[0],[1],[1],[0]] , np.float32)
```

### model 1

```python
myin = Input( shape=(2,) )
dense1 = Dense( units=100,  activation='relu')# w:2x100 b:100
dense2 = Dense( units=1,  activation='sigmoid')# w:100x1 : 1
model = Sequential( [dense1, dense2])
model.compile( loss='binary_crossentropy', optimizer=Adam(0.1), metrics=['acc'] )
h = model.fit( x,y, epochs=100 )
```

### model 3

```python
x_train = np.array([[0,0],[0,1],[1,0],[1,1]], np.float32)
y_train = np.array([[0],[1],[1],[0]], np.float32)

myin = Input( shape=(2,) )
z = Dense(units=100, activation='relu')(myin) 
hx = Dense(units=1, activation='sigmoid')(z)

model = Model(inputs=myin, outputs=hx)
model.compile(loss='binary_crossentropy', optimizer=Adam(0.1), metrics=['acc'])

h = model.fit(x_train, y_train, epochs=100)
```

### model 4

```python
class MyXOR( Model):
    def __init__(self):
        super( MyXOR, self).__init__() #Model 생성자호출

        self.dense1 = Dense( units=100, activation='relu')
        self.dense2 = Dense( units=1, activation='sigmoid')

    def call(self, inputs):
        hx1 = self.dense1(inputs)
        return self.dense2(hx1)
    
model = MyXOR()
model.compile(loss='binary_crossentropy', optimizer=Adam(0.1), metrics=['acc'])
h = model.fit(x_train, y_train, epochs=100)
```

```python
### 퀴즈 : iris 데이터 모델4 딥러닝으로  구성해서 정확도를 구하시요(히든레이어2개)
```

```python
from sklearn.datasets import load_iris
from keras.utils import to_categorical
iris = load_iris()
x_train = iris['data']
y_train = to_categorical( iris['target'] )
```

```python
class MyIris( Model):
    def __init__(self):
        super( MyIris, self).__init__() #Model 생성자호출

        self.dense1 = Dense( units=32, activation='relu')
        self.dense2 = Dense( units=16, activation='relu')
        self.dense3 = Dense( units=3, activation='softmax')

    def call(self, inputs):
        hx1 = self.dense1(inputs)
        hx2 = self.dense2(hx1)
        return self.dense3(hx2)
    
model = MyIris()
model.compile(loss='categorical_crossentropy', optimizer=Adam(0.01), metrics=['acc'])
h = model.fit(x_train, y_train, epochs=500)
```

```python
from sklearn.metrics import accuracy_score
pred = model.predict( x_train).argmax(axis=1)
accuracy_score( iris['target'], pred)
```
