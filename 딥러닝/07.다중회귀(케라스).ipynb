```python
import tensorflow as tf
from keras.optimizers import Adam
from keras import Sequential
from keras.layers import Dense
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('data/test.csv', header=None)
df.columns = ['q1','q2', 'mid','final']
df
```

```python
x = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
x.shape
```

```python
# 1. hx = matmul(x, w) + b
# 2. w [3,1], b[1]
dense = Dense( units=1, input_dim=3 )
print( dense(x) ) # hx = matmul(x,w) + b
w,b = dense.get_weights()
print( w )
print( b )
```

```python
dense = Dense( units=1, input_dim=3)
model = Sequential( [dense]) # dense의  w와 b가 학습대상이다
model.compile( loss='mse', optimizer= Adam(0.1) )
model.fit( x, y, epochs=500)
```

```python
w,b = dense.get_weights()
print( w )
print( b )
```

```python
dense(x)
```

```python
model.predict( x) # 1.기본전처리. 2.dense(x) 3. 결과를  numpy로
```

```python
# q1: 73, q2:80,  mid:75
model.predict( np.array([[73,80,75]]) ) 
```
