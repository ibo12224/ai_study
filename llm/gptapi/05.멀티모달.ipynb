### 5분 퀴즈 베트남여행.mp3 텍스트를 추출하고 요약하시요.

```python
client = OpenAI()
fp = open( 'audio/베트남여행.mp3', 'rb')

transcription = client.audio.transcriptions.create( file=fp, model='whisper-1')
# print( transcription)
print( transcription.text)

completion = client.chat.completions.create( model='gpt-3.5-turbo',
            messages=[{'role':'user',
                       'content':f'Please summarize the following:{transcription.text}' } ] )
print(completion.choices[0].message.content)
```

```python
client = OpenAI()

def transcribe_audio( file_path):
    fp = open( file_path, 'rb')
    transcription = client.audio.transcriptions.create( file=fp, model='whisper-1')
    return transcription.text

def transcribe_and_summarize( trans_text):
    completion = client.chat.completions.create( model='gpt-3.5-turbo',
                messages=[{'role':'user',
                        'content':f'Please summarize the following:{trans_text}' } ] )
    return completion.choices[0].message.content

text = transcribe_audio( 'audio/베트남여행.mp3' )
sumarize_text = transcribe_and_summarize( text )
print('요약내용', sumarize_text)
```

### tts-1 :텍스트  --> 음성

```python
client = OpenAI()
response = client.audio.speech.create(input='안녕하세요 오늘 날씨가 참 덥네요',
                                      model='tts-1', voice="nova")

with open('out.mp3','wb') as fp:
    fp.write( response.content)
```

### dalle-e: 이미지생성

```python
import requests
client = OpenAI()
response = client.images.generate(prompt="a white siamese cat",
                                  model='dall-e-3', size="1024x1024",
                                  quality="standard", n=1)
image_url = response.data[0].url
image_data = requests.get(image_url).content
with open( 'genimage.jpg','wb') as fp:
    fp.write( image_data)
```
