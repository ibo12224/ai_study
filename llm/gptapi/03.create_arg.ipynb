### stream 인자

```python
client = OpenAI()
completion = client.chat.completions.create(model='gpt-3.5-turbo',
    messages=[{'role':'user','content':'인공지능에 대해 5개로 요약해줘 각 요약 번호를 붙여줘'}],
    stream=True)

for c in completion:
    # print( c )
    if c.choices[0].delta.content:
        print( c.choices[0].delta.content, end='')
```

### n :답변수

```python
client = OpenAI()
completion = client.chat.completions.create(model='gpt-3.5-turbo',
    messages=[{'role':'user','content':'빈센트 반 고흐에 대해 알려줄래'}],
    n=2)
# completion
for choice in completion.choices:
    print( choice.message.content)
    print('-----------')
```

### max_tokens:최대 응답 토큰의 수

```python
client = OpenAI()
completion = client.chat.completions.create(model='gpt-3.5-turbo',
    messages=[{'role':'user','content':'빈센트 반 고흐에 대해 알려줄래'}],
    n=2, max_tokens=200)

for choice in completion.choices:
    print( choice.message.content)
    print('-----------')
```

### temperature
- 허용값: 0~2
- 값이 클수록 창의성(자유도)이 높아진다
- 0 보수적응답 (확실한 내용을 중심으로 답변)


```python
client = OpenAI()
completion = client.chat.completions.create(model='gpt-3.5-turbo',
    messages=[{'role':'user','content':'빈센트 반 고흐에 대해 알려줄래'}],
    temperature=0)
print( completion.choices[0].message.content)
```
