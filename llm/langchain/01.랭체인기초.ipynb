```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import SystemMessage, HumanMessage, AIMessage
```

## 기본채팅

```python
gpt = ChatOpenAI( model="gpt-3.5-turbo", temperature=0 )
rst = gpt.invoke('인공지능에 대해 간단히 설명해줘')
print( type(rst))
print( rst )
print( rst.content)
```

```python
messages= [ SystemMessage(content='You are a helpful assistant'),
    HumanMessage(content="인공지능에대해 간단히 알려줄래") ]
gpt = ChatOpenAI( model="gpt-3.5-turbo", temperature=0 )
rst = gpt.invoke( messages)
print( rst.content)
```

## 대화식 채팅

```python
messages= [ SystemMessage(content='You are a helpful assistant')]
gpt = ChatOpenAI( model="gpt-3.5-turbo", temperature=0 )
messages.append( HumanMessage(content="인공지능에대해 간단히 알려줄래") )

rst = gpt.invoke( messages)
print( rst.content)
print('----------------')
messages.append( AIMessage(content=rst.content) )

messages.append( HumanMessage(content="더간단히 알려줄래") )
rst = gpt.invoke( messages)
print( rst.content)
```

## stream

```python
gpt = ChatOpenAI( model="gpt-3.5-turbo", temperature=0 )
rst = gpt.stream('인공지능에 대해 5개로 요약해서 설명해줘')
# print( rst )
for chunk in rst:
    print( chunk.content, end="")
```

## generate

```python
messages_batch= [
    [HumanMessage(content="인공지능에 대해 간단히 알려줄래")],
    [HumanMessage(content="파이썬에 대해 알려줄래")]
]
gpt = ChatOpenAI( model="gpt-3.5-turbo", temperature=0 )
rst = gpt.generate( messages_batch, n=1)
print( rst.generations[0][0].text )
#print( rst.generations[0][1].text )
print("=====")
print( rst.generations[1][0].text )
#print( rst.generations[1][1].text )
```
